```python
# Cell 2: User inputs
# You can change these values and run all cells again.

N = 20          # Number of masses
mass = 0.010    # Mass of each small segment (kg)
k = 100.0       # Spring stiffness (N/m)

# Number of harmonics to display
num_modes = 3

print("Musical Instrument Frequency Analyzer")
print("-------------------------------------")
print(f"Number of masses       : {N}")
print(f"Mass of each segment   : {mass} kg")
print(f"Spring stiffness       : {k} N/m")
print(f"Harmonics to display   : {num_modes}")# Cell 3: Construct the stiffness matrix K

# Start with an N x N matrix of zeros
K = np.zeros((N, N))

# Fill the diagonal with 2k
for i in range(N):
    K[i, i] = 2 * k

# Fill the upper and lower diagonals with -k
for i in range(N - 1):
    K[i, i + 1] = -k
    K[i + 1, i] = -k

print("Stiffness matrix K:")
print(K)# Cell 4: Display the matrix symbolically

# Use a small matrix for easy visual understanding
display_N = min(N, 6)

K_small = sp.zeros(display_N, display_N)

for i in range(display_N):
    K_small[i, i] = 2 * sp.Symbol('k')

for i in range(display_N - 1):
    K_small[i, i + 1] = -sp.Symbol('k')
    K_small[i + 1, i] = -sp.Symbol('k')

print(f"First {display_N} x {display_N} part of the stiffness matrix:")
sp.pprint(K_small)
```

```python
# Cell 5: Construct the mass matrix

M = mass * np.eye(N)

print("Mass matrix M:")
print(M)
```

```python
# Cell 6: Calculate eigenvalues and eigenvectors

# M^(-1) K gives the matrix whose eigenvalues are omega^2
A = np.linalg.inv(M) @ K

# Since A is symmetric, eigh() is more accurate for this problem
eigenvalues, eigenvectors = np.linalg.eigh(A)

# Eigenvalues = omega^2
omega_squared = eigenvalues

# Angular frequencies
omega = np.sqrt(omega_squared)

# Ordinary frequencies in Hz
frequencies = omega / (2 * np.pi)

print("Eigenvalues (omega^2):")
print(omega_squared)

print("\nAngular frequencies (rad/s):")
print(omega)

print("\nFrequencies (Hz):")
print(frequencies)
```

```python
# Cell 7: Normalize eigenvectors

# eigh() already returns eigenvalues in ascending order.
# Each column of eigenvectors corresponds to one vibration mode.

modes = eigenvectors.copy()

# Normalize every mode so its largest displacement is 1
for i in range(N):
    modes[:, i] = modes[:, i] / np.max(np.abs(modes[:, i]))

print("First three vibration modes:")
print(modes[:, :num_modes])
```

```python
# Cell 8: Readable results table

print("\nFIRST 3 HARMONICS")
print("=" * 60)

print(f"{'Mode':<10}{'Eigenvalue (ω²)':<20}{'ω (rad/s)':<20}{'Frequency (Hz)':<20}")
print("-" * 60)

for i in range(num_modes):
    print(
        f"{i+1:<10}"
        f"{omega_squared[i]:<20.4f}"
        f"{omega[i]:<20.4f}"
        f"{frequencies[i]:<20.4f}"
    )
```

```python
# Cell 9: Plot the first 3 vibration modes

positions = np.arange(1, N + 1)

plt.figure(figsize=(10, 6))

for i in range(num_modes):
    plt.plot(
        positions,
        modes[:, i],
        marker='o',
        linewidth=2,
        label=f"Mode {i+1}  ({frequencies[i]:.2f} Hz)"
    )

plt.axhline(0, linewidth=1)
plt.xlabel("Mass Position Along String")
plt.ylabel("Relative Displacement")
plt.title("First 3 Vibration Modes of the Guitar String")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
# Cell 10: Show each harmonic separately

fig, axes = plt.subplots(num_modes, 1, figsize=(10, 9))

# Handle the case where only one mode is requested
if num_modes == 1:
    axes = [axes]

for i in range(num_modes):
    axes[i].plot(
        positions,
        modes[:, i],
        marker='o',
        linewidth=2
    )

    axes[i].axhline(0, linewidth=1)
    axes[i].set_title(
        f"Harmonic {i+1}: Frequency = {frequencies[i]:.2f} Hz"
    )
    axes[i].set_xlabel("Position Along String")
    axes[i].set_ylabel("Displacement")
    axes[i].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Cell 11: Compare the first three harmonic frequencies

harmonic_numbers = np.arange(1, num_modes + 1)

plt.figure(figsize=(8, 5))

plt.bar(
    harmonic_numbers,
    frequencies[:num_modes]
)

plt.xlabel("Harmonic Number")
plt.ylabel("Frequency (Hz)")
plt.title("First 3 Harmonic Frequencies")

plt.xticks(harmonic_numbers)
plt.grid(axis='y', alpha=0.3)

plt.show()
```

```python
# Cell 12: Compare numerical frequencies with ideal harmonic ratios

fundamental = frequencies[0]

print("HARMONIC RATIO CHECK")
print("=" * 50)

print(f"{'Mode':<10}{'Numerical f':<20}{'f/f1':<15}{'Ideal ratio':<15}")
print("-" * 50)

for i in range(num_modes):
    ratio = frequencies[i] / fundamental
    ideal_ratio = i + 1

    print(
        f"{i+1:<10}"
        f"{frequencies[i]:<20.4f}"
        f"{ratio:<15.4f}"
        f"{ideal_ratio:<15}"
    )
```

```python
# Cell 13: Small benchmark test

benchmark_N = 5
benchmark_mass = 0.01      # kg
benchmark_k = 100.0        # N/m

print("BENCHMARK DATASET")
print("=" * 40)
print(f"Masses             : {benchmark_N}")
print(f"Mass per segment   : {benchmark_mass} kg")
print(f"Spring stiffness   : {benchmark_k} N/m")

# Build benchmark stiffness matrix
K_benchmark = np.zeros((benchmark_N, benchmark_N))

for i in range(benchmark_N):
    K_benchmark[i, i] = 2 * benchmark_k

for i in range(benchmark_N - 1):
    K_benchmark[i, i + 1] = -benchmark_k
    K_benchmark[i + 1, i] = -benchmark_k

# Mass matrix
M_benchmark = benchmark_mass * np.eye(benchmark_N)

# Solve
A_benchmark = np.linalg.inv(M_benchmark) @ K_benchmark

benchmark_eigenvalues, benchmark_modes = np.linalg.eigh(A_benchmark)

benchmark_omega = np.sqrt(benchmark_eigenvalues)
benchmark_frequencies = benchmark_omega / (2 * np.pi)

print("\nBenchmark frequencies:")
for i, f in enumerate(benchmark_frequencies, start=1):
    print(f"Mode {i}: {f:.4f} Hz")
```

```python
# Cell 14: Basic correctness check

# For this model, all eigenvalues should be positive.
assert np.all(benchmark_eigenvalues > 0), "Eigenvalues should be positive."

# Frequencies should be in ascending order.
assert np.all(np.diff(benchmark_frequencies) > 0), \
    "Frequencies should increase with mode number."

print("✓ Benchmark test passed.")
print("✓ All eigenvalues are positive.")
print("✓ Frequencies are correctly ordered.")
```

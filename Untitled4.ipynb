```python
# ============================================================
# 🎸 MUSICAL INSTRUMENT FREQUENCY ANALYZER
# FRONT-END ARTIFACT / USER INTERFACE
# ============================================================

# Install/import widgets
!pip -q install ipywidgets

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML


# ============================================================
# 1. TITLE / PROJECT HEADER
# ============================================================

display(HTML("""
<div style="
    background: linear-gradient(135deg, #1e3c72, #2a5298);
    color: white;
    padding: 25px;
    border-radius: 15px;
    text-align: center;
    margin-bottom: 20px;
">
    <h1>🎸 Musical Instrument Frequency Analyzer</h1>
    <h3>B.Tech Mathematics Mini Project — B4</h3>
    <p>Eigenvalues • Eigenvectors • Vibration Modes • Harmonics</p>
</div>
"""))


# ============================================================
# 2. INPUT WIDGETS
# ============================================================

N_input = widgets.IntSlider(
    value=20,
    min=2,
    max=50,
    step=1,
    description='Masses (N):',
    style={'description_width': 'initial'}
)

mass_input = widgets.FloatText(
    value=0.010,
    description='Mass (kg):',
    style={'description_width': 'initial'}
)

k_input = widgets.FloatText(
    value=100.0,
    description='Spring k (N/m):',
    style={'description_width': 'initial'}
)

modes_input = widgets.IntSlider(
    value=3,
    min=1,
    max=10,
    step=1,
    description='Modes:',
    style={'description_width': 'initial'}
)

run_button = widgets.Button(
    description='▶ Run Analyzer',
    button_style='success',
    tooltip='Calculate frequencies and vibration modes'
)

output = widgets.Output()


# ============================================================
# 3. DISPLAY INPUT SECTION
# ============================================================

display(HTML("""
<h2>⚙️ Model Parameters</h2>
<p>
The guitar string is modeled as identical masses connected by
identical springs with fixed ends.
</p>
"""))

display(
    widgets.VBox([
        N_input,
        mass_input,
        k_input,
        modes_input,
        run_button
    ])
)

display(output)


# ============================================================
# 4. ANALYZER FUNCTION
# ============================================================

def run_analyzer(button):

    with output:

        clear_output(wait=True)

        # -----------------------------------------
        # Read user inputs
        # -----------------------------------------

        N = N_input.value
        mass = mass_input.value
        k = k_input.value
        num_modes = modes_input.value

        # -----------------------------------------
        # Validate inputs
        # -----------------------------------------

        if N < 2:
            print("❌ Number of masses must be at least 2.")
            return

        if mass <= 0:
            print("❌ Mass must be greater than zero.")
            return

        if k <= 0:
            print("❌ Spring stiffness must be greater than zero.")
            return

        if num_modes > N:
            print("❌ Number of modes cannot be greater than N.")
            return

        # -----------------------------------------
        # Mass Matrix
        # -----------------------------------------

        M = mass * np.eye(N)

        # -----------------------------------------
        # Stiffness Matrix
        # -----------------------------------------

        K = np.zeros((N, N))

        for i in range(N):
            K[i, i] = 2 * k

        for i in range(N - 1):
            K[i, i + 1] = -k
            K[i + 1, i] = -k

        # -----------------------------------------
        # Eigenvalue Problem
        #
        # K v = ω² M v
        #
        # A = M⁻¹K
        # -----------------------------------------

        A = np.linalg.inv(M) @ K

        eigenvalues, eigenvectors = np.linalg.eigh(A)

        # Eigenvalues = ω²
        omega_squared = eigenvalues

        # Angular frequencies
        omega = np.sqrt(omega_squared)

        # Frequency in Hz
        frequencies = omega / (2 * np.pi)

        # -----------------------------------------
        # Normalize eigenvectors
        # -----------------------------------------

        modes = eigenvectors.copy()

        for i in range(N):
            maximum = np.max(np.abs(modes[:, i]))

            if maximum != 0:
                modes[:, i] /= maximum

        # -----------------------------------------
        # Results Header
        # -----------------------------------------

        display(HTML("""
        <h2>📊 Analysis Results</h2>
        """))

        # -----------------------------------------
        # Result Table
        # -----------------------------------------

        rows = ""

        for i in range(num_modes):

            rows += f"""
            <tr>
                <td>{i + 1}</td>
                <td>{omega_squared[i]:.4f}</td>
                <td>{omega[i]:.4f}</td>
                <td>{frequencies[i]:.4f}</td>
            </tr>
            """

        table = f"""
        <table style="
            border-collapse: collapse;
            width: 100%;
            text-align: center;
            font-size: 15px;
        ">
            <tr style="background:#2a5298;color:white;">
                <th style="padding:10px;">Mode</th>
                <th style="padding:10px;">ω²</th>
                <th style="padding:10px;">ω (rad/s)</th>
                <th style="padding:10px;">Frequency (Hz)</th>
            </tr>

            {rows}
        </table>
        """

        display(HTML(table))

        # -----------------------------------------
        # Key Result Cards
        # -----------------------------------------

        display(HTML(f"""
        <br>

        <div style="display:flex; gap:15px;">

            <div style="
                flex:1;
                background:#e8f5e9;
                padding:15px;
                border-radius:10px;
                text-align:center;
            ">
                <h3>1st Harmonic</h3>
                <h2>{frequencies[0]:.2f} Hz</h2>
            </div>

            <div style="
                flex:1;
                background:#e3f2fd;
                padding:15px;
                border-radius:10px;
                text-align:center;
            ">
                <h3>2nd Harmonic</h3>
                <h2>{frequencies[1]:.2f} Hz</h2>
            </div>

            <div style="
                flex:1;
                background:#fff3e0;
                padding:15px;
                border-radius:10px;
                text-align:center;
            ">
                <h3>3rd Harmonic</h3>
                <h2>{frequencies[2]:.2f} Hz</h2>
            </div>

        </div>
        """))

        # -----------------------------------------
        # Vibration Modes
        # -----------------------------------------

        display(HTML("""
        <h2>〰️ Vibration Modes</h2>
        """))

        positions = np.arange(1, N + 1)

        plt.figure(figsize=(11, 6))

        for i in range(num_modes):

            plt.plot(
                positions,
                modes[:, i],
                marker='o',
                linewidth=2,
                label=f"Mode {i+1} — {frequencies[i]:.2f} Hz"
            )

        plt.axhline(0, linewidth=1)

        plt.xlabel("Mass Position Along String")
        plt.ylabel("Relative Displacement")

        plt.title(
            "Vibration Modes from Eigenvectors",
            fontsize=16
        )

        plt.grid(True, alpha=0.3)
        plt.legend()

        plt.show()

        # -----------------------------------------
        # Frequency Spectrum
        # -----------------------------------------

        display(HTML("""
        <h2>📈 Frequency Spectrum</h2>
        """))

        spectrum_modes = min(10, N)

        plt.figure(figsize=(10, 5))

        plt.bar(
            np.arange(1, spectrum_modes + 1),
            frequencies[:spectrum_modes]
        )

        plt.xlabel("Mode Number")
        plt.ylabel("Frequency (Hz)")
        plt.title("Natural Frequency Spectrum")

        plt.xticks(
            np.arange(1, spectrum_modes + 1)
        )

        plt.grid(axis='y', alpha=0.3)

        plt.show()

        # -----------------------------------------
        # Mathematical Explanation
        # -----------------------------------------

        display(HTML(f"""
        <div style="
            background:#f5f5f5;
            padding:20px;
            border-radius:12px;
            margin-top:20px;
        ">

        <h2>🧮 Mathematical Interpretation</h2>

        <p>
        The stiffness matrix <b>K</b> and mass matrix <b>M</b>
        satisfy the eigenvalue equation:
        </p>

        <h3 style="text-align:center;">
        K v = ω² M v
        </h3>

        <p>
        The calculated eigenvalues represent <b>ω²</b>,
        while the eigenvectors represent the corresponding
        <b>vibration modes</b>.
        </p>

        <p>
        The fundamental frequency obtained is:
        <b>{frequencies[0]:.4f} Hz</b>
        </p>

        </div>
        """))


# ============================================================
# 5. CONNECT BUTTON TO FUNCTION
# ============================================================

run_button.on_click(run_analyzer)
```

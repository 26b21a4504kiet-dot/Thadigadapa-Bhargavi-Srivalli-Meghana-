```python
# ============================================================
# 🎸 MUSICAL INSTRUMENT FREQUENCY ANALYZER (B4)
# B.Tech Mathematics Mini Project
# ============================================================
#
# Mathematical Concept:
# Eigenvalues of a tridiagonal stiffness matrix
# and vibration modes of a mass-spring system.
#
# Model:
# A guitar string is approximated as N identical masses
# connected by identical springs, with both ends fixed.
#
# Mathematical equation:
#       K v = ω² M v
#
# Eigenvalues = ω²
# Eigenvectors = vibration modes
# Frequency = ω / (2π)
# ============================================================


# -------------------------------
# 1. IMPORT LIBRARIES
# -------------------------------

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

# Make numerical output easier to read
np.set_printoptions(precision=4, suppress=True)


# -------------------------------
# 2. USER INPUTS
# -------------------------------

# Number of masses representing the string
N = 20

# Mass of each small segment of the string (kg)
mass = 0.010

# Spring stiffness (N/m)
k = 100.0

# Number of vibration modes/harmonics to display
num_modes = 3


# -------------------------------
# 3. CHECK INPUTS
# -------------------------------

if N < 2:
    raise ValueError("N must be at least 2.")

if mass <= 0:
    raise ValueError("Mass must be greater than zero.")

if k <= 0:
    raise ValueError("Spring stiffness must be greater than zero.")

if num_modes < 1 or num_modes > N:
    raise ValueError("num_modes must be between 1 and N.")


# -------------------------------
# 4. DISPLAY PROJECT INFORMATION
# -------------------------------

print("=" * 65)
print("       🎸 MUSICAL INSTRUMENT FREQUENCY ANALYZER")
print("=" * 65)

print("\nProject Code : B4")
print("Model        : Fixed-end mass-spring guitar string")

print("\nINPUT PARAMETERS")
print("-" * 65)
print(f"Number of masses       : {N}")
print(f"Mass of each segment   : {mass} kg")
print(f"Spring stiffness       : {k} N/m")
print(f"Modes to display       : {num_modes}")


# -------------------------------
# 5. MODEL ASSUMPTION
# -------------------------------

print("\nMODEL ASSUMPTION")
print("-" * 65)

print("""
The guitar string is approximated as a one-dimensional system
of identical masses connected by identical springs.

Both ends are assumed to be fixed.

For small vibrations:

        M x'' + K x = 0

For a vibration mode:

        K v = ω² M v

Therefore:
        Eigenvalue = ω²
        Eigenvector = vibration mode
        Frequency f = ω / (2π)
""")


# -------------------------------
# 6. CREATE MASS MATRIX
# -------------------------------

# For identical masses:
#
#       M = mI
#
# where I is the identity matrix.

M = mass * np.eye(N)

print("Mass matrix M created successfully.")
print(f"Matrix size: {N} × {N}")


# -------------------------------
# 7. CREATE STIFFNESS MATRIX
# -------------------------------

# For fixed ends, the stiffness matrix is:
#
# K = k * [
#       2  -1   0   0  ...
#      -1   2  -1   0  ...
#       0  -1   2  -1  ...
#       .   .   .   .  ...
# ]

K = np.zeros((N, N))

# Main diagonal = 2k
for i in range(N):
    K[i, i] = 2 * k

# Upper and lower diagonals = -k
for i in range(N - 1):
    K[i, i + 1] = -k
    K[i + 1, i] = -k


print("Stiffness matrix K created successfully.")
print(f"Matrix size: {N} × {N}")


# -------------------------------
# 8. DISPLAY A SMALL PART OF K
# -------------------------------

print("\nStiffness Matrix K (first 6 × 6 section):")
print("-" * 65)

display_size = min(6, N)

print(K[:display_size, :display_size])


# -------------------------------
# 9. CREATE SYSTEM MATRIX
# -------------------------------

# From:
#
#       K v = ω² M v
#
# Multiply by M⁻¹:
#
#       M⁻¹K v = ω² v
#
# Therefore the eigenvalues of M⁻¹K are ω².

A = np.linalg.inv(M) @ K

print("\nSystem matrix A = M⁻¹K created successfully.")


# -------------------------------
# 10. FIND EIGENVALUES & EIGENVECTORS
# -------------------------------

# A is symmetric, so np.linalg.eigh()
# is appropriate and numerically stable.

eigenvalues, eigenvectors = np.linalg.eigh(A)

# Eigenvalues = ω²
omega_squared = eigenvalues

# Angular frequencies
omega = np.sqrt(omega_squared)

# Frequency in Hertz
frequencies = omega / (2 * np.pi)


# -------------------------------
# 11. NORMALIZE VIBRATION MODES
# -------------------------------

modes = eigenvectors.copy()

for i in range(N):
    maximum = np.max(np.abs(modes[:, i]))

    if maximum != 0:
        modes[:, i] /= maximum


# -------------------------------
# 12. DISPLAY FIRST 3 RESULTS
# -------------------------------

print("\nFIRST THREE HARMONICS")
print("=" * 80)

print(
    f"{'Mode':<10}"
    f"{'Eigenvalue (ω²)':<22}"
    f"{'ω (rad/s)':<18}"
    f"{'Frequency (Hz)':<18}"
)

print("-" * 80)

for i in range(num_modes):
    print(
        f"{i + 1:<10}"
        f"{omega_squared[i]:<22.4f}"
        f"{omega[i]:<18.4f}"
        f"{frequencies[i]:<18.4f}"
    )


# -------------------------------
# 13. BASIC RESULT
# -------------------------------

print("\n" + "=" * 65)
print("PROJECT INITIALIZATION COMPLETED ✓")
print("=" * 65)

print(f"\nFundamental frequency = {frequencies[0]:.4f} Hz")
print(f"Second harmonic       = {frequencies[1]:.4f} Hz")
print(f"Third harmonic        = {frequencies[2]:.4f} Hz")

print("\nNext step:")
print("Visualize the eigenvectors as vibration modes.")
```
